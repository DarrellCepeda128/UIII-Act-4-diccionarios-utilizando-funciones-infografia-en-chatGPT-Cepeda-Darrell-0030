## ejemplos de Carpintero

```python
# 1. Inicialización del diccionario vacío
diccionario_carpintero = {}

# --- FUNCIONES DEL SISTEMA ---

def listar_carpintero():
    print("\n--- LISTA ACTUAL DE CARPINTEROS ---")
    if not diccionario_carpintero:
        print("El diccionario está vacío.")
    else:
        # Formato de tabla manual ajustado para los campos de carpintería
        print(f"{'ID':<6} | {'Nombre':<15} | {'Madera Pref.':<15} | {'Maquinaria':<12} | {'Taller':<15} | {'Pedidos':<8} | {'Presupuesto':<12}")
        print("-" * 90)
        for clave, datos in diccionario_carpintero.items():
            print(f"{clave:<6} | {datos['Nombre']:<15} | {datos['Madera Pref.']:<15} | {datos['Maquinaria']:<12} | {datos['Taller']:<15} | {datos['Pedidos']:<8} | ${datos['Presupuesto']:<.2f}")
    print("-" * 90 + "\n")

def agregar_carpintero():
    id_c = int(input("Ingrese el ID del carpintero: "))
    nombre = input("Ingrese el nombre: ")
    madera = input("Ingrese el tipo de madera preferida: ")
    maquinaria = input("¿Tiene maquinaria propia? (Sí/No): ")
    catalogo = input("Ingrese el catálogo de muebles (ej. Sillas, Mesas): ")
    taller = input("Ingrese el nombre o ubicación del taller: ")
    pedidos = int(input("Ingrese la cantidad de pedidos pendientes: "))
    presupuesto = float(input("Ingrese el costo estimado de presupuesto ($): "))

    diccionario_carpintero[id_c] = {
        "Nombre": nombre,
        "Madera Pref.": madera,
        "Maquinaria": maquinaria,
        "Catálogo": catalogo,
        "Taller": taller,
        "Pedidos": pedidos,
        "Presupuesto": presupuesto
    }
    print("\n✅ Carpintero agregado con éxito.")
    listar_carpintero()

def buscar_carpintero():
    id_c = int(input("Ingrese el ID del carpintero a buscar: "))
    if id_c in diccionario_carpintero:
        datos = diccionario_carpintero[id_c]
        print("\n🔍 RESULTADO DE BÚSQUEDA:")
        print(f"{'ID':<6} | {'Nombre':<15} | {'Madera Pref.':<15} | {'Maquinaria':<12} | {'Taller':<15} | {'Pedidos':<8} | {'Presupuesto':<12}")
        print("-" * 90)
        print(f"{id_c:<6} | {datos['Nombre']:<15} | {datos['Madera Pref.']:<15} | {datos['Maquinaria']:<12} | {datos['Taller']:<15} | {datos['Pedidos']:<8} | ${datos['Presupuesto']:<.2f}")
        print(f"📁 Catálogo de muebles: {datos['Catálogo']}")
    else:
        print("\n❌ No se encontró ningún carpintero con ese ID.")
    listar_carpintero()

def borrar_carpintero():
    id_c = int(input("Ingrese el ID del carpintero a eliminar: "))
    if id_c in diccionario_carpintero:
        del diccionario_carpintero[id_c]
        print(f"\n🗑️ Carpintero con ID {id_c} eliminado.")
    else:
        print("\n❌ El ID no existe.")
    listar_carpintero()

def actualizar_carpintero():
    id_c = int(input("Ingrese el ID del carpintero a actualizar: "))
    if id_c in diccionario_carpintero:
        print("Deje en blanco si no desea cambiar el valor.")
        nombre = input(f"Nuevo nombre ({diccionario_carpintero[id_c]['Nombre']}): ")
        madera = input(f"Nueva madera preferida ({diccionario_carpintero[id_c]['Madera Pref.']}): ")
        maquinaria = input(f"¿Maquinaria propia? ({diccionario_carpintero[id_c]['Maquinaria']}): ")
        catalogo = input(f"Nuevo catálogo de muebles ({diccionario_carpintero[id_c]['Catálogo']}): ")
        taller = input(f"Nuevo taller ({diccionario_carpintero[id_c]['Taller']}): ")
        pedidos_input = input(f"Nuevos pedidos pendientes ({diccionario_carpintero[id_c]['Pedidos']}): ")
        presupuesto_input = input(f"Nuevo costo de presupuesto ({diccionario_carpintero[id_c]['Presupuesto']}): ")

        if nombre: diccionario_carpintero[id_c]['Nombre'] = nombre
        if madera: diccionario_carpintero[id_c]['Madera Pref.'] = madera
        if maquinaria: diccionario_carpintero[id_c]['Maquinaria'] = maquinaria
        if catalogo: diccionario_carpintero[id_c]['Catálogo'] = catalogo
        if taller: diccionario_carpintero[id_c]['Taller'] = taller
        if pedidos_input: diccionario_carpintero[id_c]['Pedidos'] = int(pedidos_input)
        if presupuesto_input: diccionario_carpintero[id_c]['Presupuesto'] = float(presupuesto_input)

        print("\n🔄 Datos actualizados.")
    else:
        print("\n❌ ID no encontrado.")
    listar_carpintero()

# --- MENÚ PRINCIPAL ---

while True:
    print("*******************************")
    print("  SISTEMA DE CONTROL DE TALLER ")
    print("*******************************")
    print("1. Agregar Carpintero")
    print("2. Listar Carpinteros")
    print("3. Buscar Carpintero")
    print("4. Borrar Carpintero")
    print("5. Actualizar Carpintero")
    print("6. Salir")

    opcion = input("\nSeleccione una opción: ")

    if opcion == '1':
        agregar_carpintero()
    elif opcion == '2':
        listar_carpintero()
    elif opcion == '3':
        buscar_carpintero()
    elif opcion == '4':
        borrar_carpintero()
    elif opcion == '5':
        actualizar_carpintero()
    elif opcion == '6':
        print("Saliendo del sistema... ¡Buen trabajo en el taller!")
        break
    else:
        print("Opción no válida, intente de nuevo.")
print("Programa elaborado por Cepeda Darrell NL = 10")
```
